````python
def genkeys(detectors:list):
    modules = expand_modules(detectors, dict_module_to_uniqch)
    keys = []
    for m in modules:
        if m[:2] in ["M1", "M2", "M3", "M4"]:#, "M6" ]:
            continue
        # if m in ["C3(2)", "C5(1)", "C5(2)", "C6(1)", "C6(2)"]:
        #     continue
        # if m in ["P05", "P06", "P15","P16", "P17", "P19", "P20", "P21", "P26", "P28", "P31"]:
        #     continue
        # if m not in ["P41", "P38", "P37", "P29", "P34", "P32", "P17", "P25", "P35"]:
        # if m not in ["P12", "P06", "P14", "P35"]:
        if m.startswith('P') and m not in ["P41", "P38", "P37", "P29", "P34", "P32", "P17", "P25", "P35"]:#, "P12", "P06", "P14"]:
            continue
        uch:UniqueChannel = dict_module_to_uniqch[m]
        ep, ch = uch.endpoint, uch.channel
        keys.append((ep,ch))

    return keys
````

````python
files = glob(f"{data_folder}/*.txt")
print(len(files), 'files found')
pattern = "ppm_([\d,\.]+)_ep_(\d+)_ch_(\d+).txt"
datasetfull = {}
results_per_ppm = {}
for file in files:
    fname = os.path.basename(file)
    match = re.match(pattern, fname)
    ppm = match.group(1)
    # if float(ppm) < 0.01:
    #     continue
    ep = int(match.group(2))
    ch = int(match.group(3))
    wf = np.loadtxt(file, dtype=np.float32)
    wf = wf - np.mean(wf[:40])
    datasetfull[ppm] = datasetfull.get(ppm, {})
    datasetfull[ppm][(ep, ch)] = wf

datasetfull = { ppm : v for ppm, v in sorted(datasetfull.items(), key=lambda x: float(x[0])) }
````

````python
trueppm = {}
for ppm in datasetfull.keys():
    if ppm == '0.00':
        trueppm[ppm] = 0.0
        continue
    imatch = np.argmin(abs(float(ppm) - injppm))
    trueppm[ppm]  = injppm[imatch]
trueppm
````

````python
# %% [markdown]
# ## Estimate per-channel uncertainty (sigma) from pre-pulse baseline noise
#
# Works on the deconvolved waveforms directly (whatever units/processing state
# `gf.datasets[key]` is currently in). Also checks lag-1 autocorrelation to flag
# whether deconvolution introduced sample-to-sample correlation (relevant if you
# ever move to a full covariance-matrix chi2 instead of diagonal).

# %%
import numpy as np
import matplotlib.pyplot as plt

def true_baseline_rms(gf, key, before_ns=100, min_points=20):
    """
    Estimate noise sigma from the pre-pulse baseline (before t0 - before_ns).
    Returns np.nan if not enough baseline points are available.
    """
    x = gf.x
    t0 = gf.fit_results.modules[key].t0.value
    y = gf.datasets[key]
    mask = x < (t0 - before_ns)
    baseline = y[mask]
    if len(baseline) < min_points:
        return np.nan
    return np.std(baseline)

def lag1_autocorr(y):
    """Lag-1 autocorrelation coefficient of a 1D array (noise correlation check)."""
    y = y - np.mean(y)
    num = np.sum(y[:-1] * y[1:])
    den = np.sum(y**2)
    return num / den if den > 0 else np.nan

# %%
def estimate_all_sigmas(gf, before_ns=100):
    """
    Loop over all channels, compute baseline RMS and lag-1 autocorrelation.
    Returns a dict {key: {'rms':..., 'autocorr':..., 'n_baseline':...}}
    """
    results = {}
    for key in gf.datasets:
        x = gf.x
        t0 = gf.fit_results.modules[key].t0.value
        y = gf.datasets[key]
        mask = x < (t0 - before_ns)
        baseline = y[mask]

        rms = np.std(baseline) if len(baseline) >= 20 else np.nan
        ac  = lag1_autocorr(baseline) if len(baseline) >= 20 else np.nan

        results[key] = {"rms": rms, "autocorr": ac, "n_baseline": len(baseline)}
    return results

results = estimate_all_sigmas(gf, before_ns=200)

# Pretty print
print(f"{'Channel':<12}{'n_baseline':>12}{'RMS (sigma)':>14}{'lag-1 autocorr':>18}")
for key, r in results.items():
    ep, ch = key
    label = f"{ep}-{ch}"
    print(f"{label:<12}{r['n_baseline']:>12}{r['rms']:>14.4f}{r['autocorr']:>18.3f}")

# %% [markdown]
# ### Interpretation
# - `RMS (sigma)` -> use this as your per-channel sigma (or average/median across
#   channels if you want ONE flat value for all detectors).
# - `lag-1 autocorr` -> if this is well above ~0.2-0.3 in magnitude, the noise is
#   meaningfully correlated sample-to-sample (expected after deconvolution).
#   The diagonal chi2 (sum of independent (y-f)^2/sigma^2) will then slightly
#   misstate your effective dof, but using the empirical RMS as sigma is still
#   the right central estimate -- correlation mainly affects how much you should
#   trust chi2/ndof as a p-value, not the sigma value itself.

# %%
# Aggregate stats -- useful if you want one shared sigma across all channels
rms_values = np.array([r["rms"] for r in results.values() if not np.isnan(r["rms"])])

print(f"\nAcross {len(rms_values)} channels:")
print(f"  mean sigma   = {rms_values.mean():.4f}")
print(f"  median sigma = {np.median(rms_values):.4f}")
print(f"  std of sigma across channels = {rms_values.std():.4f}")
print(f"  min / max    = {rms_values.min():.4f} / {rms_values.max():.4f}")

# %%
# Quick visual: sigma per channel, sorted
sorted_items = sorted(results.items(), key=lambda kv: kv[1]["rms"])
labels = [f"{k[0]}-{k[1]}" for k, _ in sorted_items]
vals   = [v["rms"] for _, v in sorted_items]

plt.figure(figsize=(10, 4))
plt.bar(labels, vals)
plt.xticks(rotation=90)
plt.ylabel("Baseline RMS (sigma)")
plt.axhline(np.median(vals), color="red", ls="--", label=f"median = {np.median(vals):.3f}")
plt.legend()
plt.tight_layout()
plt.show()

# %% [markdown]
# ### Using the result
# Option A -- per-channel sigma (more correct, since spread above shows sigma
# does vary channel to channel):
# ```python
# gf.errors = {key: np.full_like(y, results[key]["rms"]**2)
#              for key, y in datasets.items()}
# ```
#
# Option B -- one flat sigma for all channels (simpler, matches your stated
# preference for no per-channel/time-dependent uncertainty):
# ```python
# sigma_flat = np.median(rms_values)   # or mean(rms_values)
# gf.errors = {key: np.full_like(y, sigma_flat**2) for key, y in datasets.items()}
# ```
#
# Either way, rerun `minimize()` (or just re-evaluate chi2/reduced_chi2) with the
# new `errors` dict and check `m.fmin.reduced_chi2` again.
````

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.odr import ODR, Model, RealData
from dataclasses import dataclass
from typing import Optional, Callable
import pandas as pd


@dataclass
class ODRResult:
    slope:         float
    intercept:     float
    err_slope:     float
    err_intercept: float

    def __repr__(self):
        return (f"slope = {self.slope:.4f} ± {self.err_slope:.4f}, "
                f"intercept = {self.intercept:.4f} ± {self.err_intercept:.4f}")


def fit_inverse_tau(ppm, tau, err_tau, ppm_err=0.05, beta0=[1, 0]) -> ODRResult:
    """ODR linear fit on 1/tau vs ppm. Returns ODRResult."""
    data = RealData(ppm, 1/tau, sx=ppm_err, sy=err_tau/tau**2)
    out  = ODR(data, Model(lambda B, x: B[0]*x + B[1]), beta0=beta0).run()
    return ODRResult(
        slope         = out.beta[0],
        intercept     = out.beta[1],
        err_slope     = out.sd_beta[0],
        err_intercept = out.sd_beta[1],
    )


def plot_inverse_tau(ax, ppm, tau, err_tau, result, label, color, ppm_err=0.05, coeff=["A", "B"]):
    """Plot data points + ODR fit line on a given axis."""
    x_fit = np.linspace(ppm.min(), ppm.max(), 200)
    y_fit = result.slope * x_fit + result.intercept

    ax.errorbar(ppm, 1/tau, yerr=err_tau/tau**2, xerr=ppm_err,
                fmt='.', label=label, color=color)
    ax.plot(x_fit, y_fit, color=color,
            label=(rf'Fit {label}: '
                   rf'${result.intercept:.3f}\pm{result.err_intercept:.3f}$ + '
                   rf'${result.slope:.3f}\pm{result.err_slope:.3f} \cdot \text{{Xe [ppm]}}$'))


def run_tau_analysis(df: pd.DataFrame,
                     tau_col: str,
                     err_col: str,
                     row_filter: Optional[Callable] = None,
                     ppm_err: float = 0.05) -> tuple[ODRResult, pd.DataFrame]:
    """
    Filter df, fit 1/tau vs ppm, return (ODRResult, filtered_df).

    row_filter: any callable that takes a DataFrame and returns a filtered DataFrame, e.g.:
        row_filter = lambda df: df[df["ppm"] <= 4]
        row_filter = lambda df: df[df["ppm"].isin([1, 2, 3])]
        row_filter = lambda df: df[df["ppm"] >= 2][df["tta"] < 1200]
    """
    data = df.copy()
    if row_filter is not None:
        data = row_filter(data)

    result = fit_inverse_tau(data["ppm"], data[tau_col], data[err_col], ppm_err)
    return result, data


# --- Usage ---
result_tx, data_tx = run_tau_analysis(
    df, "ttx", "errttx",
    row_filter = lambda df: df[df["ppm"] > 0.1],
)
result_ta, data_ta = run_tau_analysis(
    df, "tta", "errtta",
    row_filter = lambda df: df[~df["ppm"].between(0.008, 0.02)],
)

fig, ax = plt.subplots(figsize=(8, 6))

plot_inverse_tau(ax, data_ta["ppm"], data_ta["tta"], data_ta["errtta"],
                 result_ta, label=r'$\tau_{TA}$', color='tab:orange')

plot_inverse_tau(ax, data_tx["ppm"], data_tx["ttx"], data_tx["errttx"],
                 result_tx, label=r'$\tau_{TX}$', color='tab:blue',
                 coeff=["C", "D"])

ax.set_ylabel(r'$1\,/\,\tau\;[\mu s^{-1}]$')
ax.set_xlabel(r'Xe [ppm]')
ax.legend(loc='upper left')
fig.tight_layout()
plt.savefig('graphs_global/linear_global.png', dpi=300)


plt.show()
# print(f"TA: {result_ta}")
# print(f"TX: {result_tx}")
````
